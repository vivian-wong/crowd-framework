```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, HBox, VBox, jslink
from IPython.display import display
import os

from st_dif.data_utils import get_pyg_temporal_dataset
from st_dif.vis_utils import show_crowd_table

# for relative imports
os.chdir('..') 
print(os.getcwd()) # should print /your_local_dir/crowd-framework
```

```python
DATASET      = "Stadium"        # or "SEQ", "GCS", etc.
FORECAST_NS  = 20               # your forecasting_horizon
NODE_ORDER   = list(range(g.X.shape[0]))   # number of PARs(nodes)

print("Loading dataset...")
_, g = get_pyg_temporal_dataset(DATASET, FORECAST_NS)
print("Dataset loaded successfully.")

# Determine the maximum possible starting time for the slider
max_t_start = g.X.shape[2] - 1 # Total number of timesteps minus one
```

```python
# vivian changed this to interactive table. 
def interactive_table_plot(t_start, n_rows):
    """
    This function is called by the widgets to redraw the plot.
    """
    # Ensure the window does not go out of bounds
    if t_start + n_rows > g.X.shape[2]:
        n_rows = g.X.shape[2] - t_start
        print(f"Warning: n_rows adjusted to {n_rows} to fit data.")

    # Generate the plot using your existing function
    fig = show_crowd_table(
        g,
        t_start=t_start,
        n_rows=n_rows,
        dataset_name=DATASET,
        node_order=NODE_ORDER,
    )
    
    # Set a dynamic title
    title = f"{DATASET}: Timesteps {t_start}–{t_start + n_rows - 1}"
    plt.suptitle(title, y=1.02, fontsize=14, weight="bold")
    
    # Display the plot in the notebook
    plt.show()
    # Close the figure to prevent it from displaying twice in some environments
    if fig:
        plt.close()

# --- Create the widgets ---

# 1. Slider for the starting timestep (t_start)
t_slider = widgets.IntSlider(
    value=0,
    min=0,
    max=max_t_start,
    step=1,
    description='Start Time (t₀):',
    continuous_update=False, # Only update when the user releases the slider
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='500px')
)

# 2. Play widget to animate the slider
play_button = widgets.Play(
    value=0,
    min=0,
    max=max_t_start,
    step=1,
    interval=500, # Animation speed in milliseconds
    description="Press play",
)

# 3. Text box for the number of rows
n_rows_box = widgets.IntText(
    value=5,
    description='Num Rows (n):',
    style={'description_width': 'initial'}
)

# Link the Play button to the slider
jslink((play_button, 'value'), (t_slider, 'value'))

# --- Create the UI Layout ---

# Arrange the play button and slider together
slider_controls = HBox([play_button, t_slider])

# Arrange all controls together
ui_controls = HBox([slider_controls, n_rows_box])

# Create the interactive output linked to the widgets
interactive_output = widgets.interactive_output(
    interactive_table_plot, 
    {'t_start': t_slider, 'n_rows': n_rows_box}
)

# --- Display the final widget ---
print("Interactive controls are ready.")
display(ui_controls, interactive_output)
```
